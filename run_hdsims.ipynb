```python
import os
import numpy as np
from hdsims import hdsims, hdsimsutils
```

# Instructions to generate HD sims

In the cell below, you **must** provide the paths to your `hd_sims_dir` and `lowres_sims_dir`.

The other options below that are set to their default values, which you may change.
- For more information about these options, you may:
  - run `run_hdsims.py --help` 
  - refer to the [example](https://github.com/CMB-HD/hdsims/blob/main/examples/example_2x2.ipynb) Jupyter notebook
  - refer to the docstrings in the modules of `hdsims` (e.g., for the `hdsims.HDSims` class)

After you run this cell, instructions will be printed out with command(s) to download the full-sky S10 sims, if necessary, and then to run the `run_hdsims.py` file. 

Once those commands have finished running, you may initialize the `HDSims` class below to plot the maps and/or their power spectra.
- **Note**: if you don't wait for the commands to finish running before trying to make the plots, everything will be run within this notebook; we don't do anything to prevent this, so proceed with caution!

```python
# you must put the paths to your `hd_sims_dir` and `lowres_sims_dir` below:
hd_sims_dir = ''
lowres_sims_dir = ''

# you may change the following (the values are set to the defaults):
ra_ctr = 6
dec_ctr = 6
width = 10
height = width
apod_width = 0.5
cmb_seed = 58
pol = True
freqs = [30, 90, 148, 219, 277, 350]
components = ['cmb', 'ksz', 'tsz', 'cib', 'radio']

calculate_power_spectra = True # whether to calculate the power spectra of the sims you generate
save_plots = True # whether to plot the maps (and power spectra, if it was calculated)

path_to_run_hdsims = os.getcwd() # path to the `run_hdsims.py` file - should be the same as this notebook


# (don't change below:)
hdsimsutils.make_run_hdsims_command(path_to_run_hdsims, hd_sims_dir, lowres_sims_dir, 
                                    ra_ctr=ra_ctr, dec_ctr=dec_ctr, width=width, height=height,
                                    apod_width=apod_width, cmb_seed=cmb_seed, pol=pol,
                                    freqs=freqs, components=components, 
                                    calculate_power_spectra=calculate_power_spectra, save_plots=save_plots)
```

# Plots

**NOTE**: if you are running `run_hdsims.py`, you should not execute the cells below until after you have finished generating the simulations (and taking their power spectra, if you set `calculate_power_spectra=True` above)

In the cell below:
- To plot the maps, set `plot_maps=True`
  - You may change which map frequencies to plot by changing the `map_freqs_to_plot` list
- To plot the power spectra, set `plot_spectra=True`.

```python
plot_maps = False
map_freqs_to_plot = [90] # which map frequencies to plot, if `plot_maps=True`

plot_spectra = False   
```

```python
# initialize the HDSims class:
simlib = hdsims.HDSims(hd_sims_dir, ra_ctr=ra_ctr, dec_ctr=dec_ctr, width=width, height=height,
                       apod_width=apod_width, cmb_seed=cmb_seed, pol=pol,
                       freqs=freqs, components=components)
```

```python
if plot_maps:
    for freq in map_freqs_to_plot:
        print(f"{freq} GHz maps:")
        simlib.plot_sim_maps(freq)
```

```python
if plot_spectra and calculate_power_spectra:
    simlib.plot_sim_spectra_comparison()
    simlib.plot_smallscale_ksz_kappa_spectra()
```
