# Reproducing Figures 3, 4, and 5 of MacInnis et. al. 2026

In this notebook, we will reproduce Figures 4 and 5 of **[TODO:LINK2PAPER]()**, which compare the power spectra of the HD sims to their lower-resolution counterparts and to theory power spectra. We include an option to reproduce Figure 3, along with instructions on how to download the maps that appear in the plot. 

We also provide instructions on how to reproduce *all* of the HD simulation products (also available on **[TODO:LINK2LAMBDA]()**) that were used in **[TODO:LINK2PAPER]()**.

---

Below, we import the python packages needed to run this notebook:

```python
import os
%matplotlib inline
from hdsims import hdsimsutils
```

---

By default, we will only reproduce Figures 4 and 5 below.
- We provide all of the power spectra used in these plots in the `hdsims` github repository. By default, these are used to make Figures 4 and 5, so you don't need to calculate anything before making those plots.


We also provide an option to reproduce Figure 3. This requires saving the 90 GHz HD sims that are shown in the plot, so it is not done by default. If you would like to reproduce Figure 3, set `plot_maps = True` in the cell below. 
- We will provide instructions on how to download the 90 GHz HD sims, which will require about 17 GB of storage. The files will be saved to the `hd_sims_dir`, defined in the next cell. By default, this will be the same directory as this notebook; you may change this to a different path.
- After downloading the files, you may run the "Figure 3" section of this notebook. **Note** that this will require loading in 17 GB of maps. Therefore, we also provide instructions on how to save a copy of Figure 3 by running a python script instead of using this notebook.


By default, we will save the plots to the current directory (i.e., the same directory as this notebook). You may choose your own file names by changing `fig3_fname`, `fig4_fname`, and `fig5_fname` in the next cell.


If you would like to reproduce *all* of the HD simulation products used in **[TODO:LINK2PAPER]()**, you can set `reproduce_sims_and_spectra = True` (default is `False`). We will print out instructions on how to do this by running the `reproduce_10x10.py` python script, which will also save the plots we reproduce below.
- In this case, you **must** provide a path to the `lowres_sims_dir`, defined in the next cell, where the full-sky [Sehgal et. al.](https://arxiv.org/abs/0908.0540) simulations will be saved.
- Note: this will also save Figures 3, 4, and 5 under your `hd_sims_dir`.

```python
plot_maps = False
reproduce_sims_and_spectra = False

hd_sims_dir = os.getcwd()
lowres_sims_dir = ''
fig3_fname = 'fig3.png'
fig4_fname = 'fig4.pdf'
fig5_fname = 'fig5.pdf'

hdsimsutils.print_instructions_for_10x10(hd_sims_dir, lowres_sims_dir=lowres_sims_dir, 
                                         plot_maps=plot_maps, reproduce_sims_and_spectra=reproduce_sims_and_spectra,
                                         fig3_fname=fig3_fname, fig4_fname=fig4_fname, fig5_fname=fig5_fname)
```

---

# Figure 4

Below, we will reproduce Figure 4, which compares the power spectra of the HD sims to either the power spectra of the S10 sims from which they were generated, or to the corresponding theory curves for the lensed CMB spectra.

If you set `reproduce_sims_and_spectra=True` above, we will use the power spectra you have calculated when making the plot. Otherwise, we will use the power spectra that is provided with `hdsims`.

```python
hd_sims_dir_for_plot = hd_sims_dir if reproduce_sims_and_spectra else None 
hdsimsutils.reproduce_sim_spectra_comparison_plot(hd_sims_dir=hd_sims_dir_for_plot, fname=fig4_fname)
```

---

# Figure 5

Below, we will reproduce Figure 5, which compares the power spectra of the HD kSZ and lensing convergence sims, and the S10 sims on the same patch of sky, to their corresponding theory curves.

If you set `reproduce_sims_and_spectra=True` above, we will use the power spectra you have calculated when making the plot. Otherwise, we will use the power spectra that is provided with `hdsims`.

```python
hd_sims_dir_for_plot = hd_sims_dir if reproduce_sims_and_spectra else None
hdsimsutils.reproduce_ksz_kappa_sim_spectra_plot(hd_sims_dir=hd_sims_dir_for_plot, fname=fig5_fname)
```

---

# Figure 3

Below, we plot the 90 GHz HD sims for each component. All except the lensing convergence sim have been convolved with the pixel window function, and they will also be convolved with the beam. Before making the plot, we will make sure the maps have been saved.

```python
if plot_maps and hdsimsutils.fig_maps_are_saved(hd_sims_dir):
    hdsimsutils.reproduce_sim_maps_plot(hd_sims_dir, freq=90, show=True, fname=fig3_fname)
```
